```python
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine

# 1. 连接 MySQL（使用 pymysql）
engine = create_engine("mysql+pymysql://root@localhost:3306/taobao_analysis",
                       connect_args={"password": "Ciallo～(∠・ω＜ )⌒☆"})

# 2. 读取数据
df = pd.read_sql("SELECT date, Daily_UV, Daily_Sales FROM daily_stats ORDER BY date", con=engine)
print("数据读取成功，共 {} 天数据".format(len(df)))
print(df)

# 3. 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用黑体
plt.rcParams['axes.unicode_minus'] = False   # 解决负号显示问题

# 4. 绘制双轴图
fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.plot(df['date'], df['Daily_UV'], 'b-o', linewidth=2, markersize=6, label='每日UV')
ax1.set_xlabel('日期', fontsize=12)
ax1.set_ylabel('独立购买用户 (UV)', color='b', fontsize=12)
ax1.tick_params(axis='y', labelcolor='b')
ax1.tick_params(axis='x', rotation=45)

ax2 = ax1.twinx()
ax2.plot(df['date'], df['Daily_Sales'], 'r-s', linewidth=2, markersize=6, label='每日订单量')
ax2.set_ylabel('订单量', color='r', fontsize=12)
ax2.tick_params(axis='y', labelcolor='r')

plt.title('每日活跃购买用户(UV)与订单量趋势', fontsize=16, fontweight='bold')
ax1.grid(True, linestyle='--', alpha=0.6)

# 5. 保存图片
plt.savefig('daily_uv_sales_trend.png', dpi=300, bbox_inches='tight')
print("✅ 图片已保存为：daily_uv_sales_trend.png")

# 6. 显示
plt.show()
```
